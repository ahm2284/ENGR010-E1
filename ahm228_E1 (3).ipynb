#### __Section 1: Calculating the total volume of the pyramid__ 

__Explanation:__ <br> For this step, we were asked to find the total volume of the pyramid after being given different measurements in meters. We were given two measurements for the based, 253.1 meters and 253.1 meters for the length and width of the base and know that multiplying them together gets the value for the base. Following that, we had to calculate the volume of the pyramid after given a value for height, 147.2 meters, and the equation 1/3 * B * H = volume of a pyramid. To do these calculations in Python, we first created a single variable for the base where we set it to the product of the given length and width of the pyramid. Since these values have decimal places, the variable base would be a float which could be shown through print(type(base)). Following that, we created a height variable and set it equal to the given height. With all the dimensions, we used the volume of a pyramid equation to multiply the values in the base and height variables by 1/3 and assigned that calculated value to the variable volume.

```python
base = 253.1 * 253.1
height = 147.2 
volume = 1/3 * base * height
print("Total pyramid volume:", volume, "cubic meters")
```

#### __Section 2: Calculating the volume of the room__ 

__Explanation:__ <br> In this section, we are asked to either do one of two things depending on the value of the calculation above. If the value of the calculation is a certain value, then the code will execute one set of code. If the value of the calculation is a different value, then the code will execute a different set of code. For the pyramid, Imhotep wants the room to be 15% of the original volume if the volume is greater than 1.5 million cubic meters, or he wants it to be 12% of the original volume if the volume is less than 1.5 million cubic meters. To show this, we used an if-elif statement. We said that if the volume was greater than 1500000, then the volume of the room would equal the volume multiplied by 0.15, and that value was assigned to a variable roomVolume and later printed. For the elif statement, we had to create it so that if the volume was less than 1.5 million cubic meters, then it would execute a different code, satisfying the second condition of the room size. We similarly set the product of the volume and, with this execution of code, 0.12 to the variable roomVolume and printed it if the condition was met.

```python
if volume > 1500000:
    roomVolume = volume * 0.15
    print("Volume of the room:", roomVolume, "cubic meters")
elif volume < 1500000:
    roomVolume = volume * 0.12
    print("Volume of the room:", roomVolume, "cubic meters")
```

#### __Section 3: Calculating the volume occupied by stones and volume of a stone__

__Explanation:__<br>For this section, we had to calculate the volume that the stone would occupy in the pyramid. To do this, we would have to calculate the difference between the original volume and the volume of the room inside the pyramid. In the code, we showed this through subtracting the variables volume and roomVolume and assigning that value to a new variable, occupiedVolume. Following that, we had to calculate the volume of a singular stone that will be used when building the pyramid. With the measurements given, we realized that the stone was cubic with a side value of 0.93 meters, so we used the code to cube 0.93 by using ** and assigning that value to a new variable, stoneVolume.  

```python
occupiedVolume = volume - roomVolume
print("Volume that the stones will occupy:", occupiedVolume, "cubic meters")
stoneVolume = 0.93 ** 3
```

#### __Section 4: Calculate the number of stones required for the pyramid__

__Explanation:__<br> In this section, we had to calculate the total amount of stones needed to complete the pyramid. With the variables calculated in the previous section, we can conclude that the total amount of stones required would be the value of the volume that the stones occupy divided by the volume of a singular stone. To code this, we created a new variable, stonesRequired, and set it equal to occupiedVolume divided by stoneVolume. Since all the variables before were floats, we realized that the value in the variable will be a float but when purchasing materials, they are usually sold in whole numbers. To fix this problem, we created a new variable, stonesRequiredFull, and assigned the stonesRequired as an integer, using int(), and added 1 to that value because the int() function rounds down and when purchasing something, it needs the next whole value up.

```python
stonesRequired = occupiedVolume / stoneVolume
stonesRequiredFull = int(stonesRequired) + 1
print("The exact amount of stones required:", stonesRequired, "stones")
print("The exact amount of stones to purchase:", stonesRequiredFull, "stones")
```

#### __Section 5: Calculate a 10% contingency to our calculated amount of stones__

__Explanation:__<br>In the final section, to be safe, we added a 10% contingency on the exact amount of stones required to take into account any unintended situations. To do this, we created a new variable, extraStonesRequired, which takes the exact amount of stones required multiplied by 1.10 to include that extra 10%. Following that, since we know that to purchase stones, we would need the next whole number amount of stones. We created a new variable, extraStonesRequiredFull, and set it equal to the integer value of extraStonesRequired plus 1 since the int() function cuts off the decimal place and no matter how large or small the decimal is, we would need to round up. Lastly, we printed our final answers, the exact amount required to build the pyramid and the amount that should be purchased to successfully build the pyramid. 

```python
extraStonesRequired = stonesRequired * 1.10
print("The pyramid will require", extraStonesRequired, "stones to build.")
extraStonesRequiredFull = int(extraStonesRequired) + 1
print("The pyramid will need to purchase", extraStonesRequiredFull, "stones to build.")
```
